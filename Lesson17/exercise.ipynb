```python
# Gensim - Create Embeddings; 
import gensim
import pandas as pd
```

```python
# Create dataframe of reviews
data = pd.read_json("reviews_Sports_and_Outdoors_5.json", lines=True)
data.head(2)
```

```python
#Let's have a look over the review text
data.reviewText[:5]
```

```python
# Let's how many reviews we have
data.shape[0]
```

```python
# Simple Preprocessing unsing gensim
t2 = data.reviewText[0]
pt2 = gensim.utils.simple_preprocess(t2)
print(pt2)
```

```python
#Let's Apply on whole Series
review_text = data.reviewText.apply(gensim.utils.simple_preprocess)
review_text[:5]
```

```python
# Moving Context Window
model = gensim.models.Word2Vec(window=10, min_count=2, workers=6)
```

```python
# Build Vocabulary
model.build_vocab(review_text, progress_per=1000)
```

```python
#Let's train the model
model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
# Let's Examine its ability to find similar words
model.wv.most_similar("punch")
```
