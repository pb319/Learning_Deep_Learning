```python
import pandas as pd
import gensim
```

```python
df = pd.read_json("/home/prasun/GitDemo/Learning_Deep_Learning/reviews_Cell_Phones_and_Accessories_5.json",lines=True)
df.head()
```

```python
# How many records??
df.shape
```

```python
# Let's check the reviewText
df.reviewText
```

```python
# Observe in-built preprocessing function
t1 = df.reviewText[0]
pt1 = gensim.utils.simple_preprocess(t1)
print(pt1[:10])
```

```python
# Let's apply it on the whole series of reviews
review_text = df['reviewText'].apply(gensim.utils.simple_preprocess)
review_text[:5]
```

```python
# Moving the Context Window
model = gensim.models.Word2Vec(window=10, min_count=2, workers=4)
```

```python
#Building the Vocabulary
model.build_vocab(review_text, progress_per=1000)
```

```python
# Checkning Model Parameters
model.epochs
```

```python
#Lets Train the Model
model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
#Let's save the model
model.save("./word2vec-amazon-cell-accessories")
```

```python
# Let's Examine its ability to find similar words
model.wv.most_similar("good")
```

```python
# Cosine Similarity Score
model.wv.similarity(w1="good", w2="great"), model.wv.similarity(w1="good", w2="bad"), model.wv.similarity(w1="good", w2="phone")
```
